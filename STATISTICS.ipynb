### **What is Statistics?**
Statistics is the science of **collecting, organizing, presenting, analyzing, and interpreting data** to make more effective decisions.

### **Why do we use Statistics?**
1. **To summarize data:** Turn a large table of numbers (like your `students` dataset) into a few understandable metrics (like average hours or score).
2. **To make decisions:** Inform business, scientific, or personal choices based on evidence rather than intuition.
3. **To find patterns:** Understand relationships (e.g., *does studying more hours lead to a higher score?*).
4. **To make predictions:** Use past data to forecast future trends.

### **The Two Main Branches of Statistics**
* **Descriptive Statistics:** Methods of organizing, summarizing, and presenting data in an informative way (e.g., Mean, Median, Mode, Standard Deviation, Histograms).
* **Inferential Statistics:** Methods used to estimate a property of a population based on a sample (e.g., Hypothesis testing, Confidence intervals).

### **Definitions: Spread, Outliers, and Relationships**

To understand data beyond basic averages (Mean and Median), we must examine its distribution using these three core concepts:

#### **1. Spread (Variability / Dispersion)**
* **Definition:** Spread describes how scattered, stretched, or clustered the data points are. It tells us how much the data varies around the center.
* **Key Metrics:**
  * **Range:** The difference between the maximum and minimum value ($90 - 12 = 78$ in our dataset).
  * **Variance & Standard Deviation ($σ$ / $s$):** Measures the average distance of each data point from the mean.
  * **Interquartile Range (IQR):** The middle 50% of the data (Q3 - Q1), which ignores extreme outliers.

#### **2. Outliers**
* **Definition:** An outlier is an extreme data point that differs significantly from the rest of the dataset. It lies unusually far from the mean or the bulk of the data.
* **Why they matter:** As we demonstrated mathematically, an outlier (like $90$) severely distorts the **mean** (dragging it up to $26.1$), whereas the **median** ($21.0$) remains a stable measure of center.

#### **3. Relationships**
* **Definition:** A relationship (or association) refers to how two or more variables behave together.
* **Key Concepts:**
  * **Correlation:** Measures the strength and direction of a linear relationship between two quantitative variables (e.g., does studying more **hours** lead to a higher exam **score**?).
  * **Positive Relationship:** Both variables increase together.
  * **Negative Relationship:** As one variable increases, the other decreases.

```python
import numpy as np
import pandas as pd

# 1. Setup the initial arrays
x = np.array([1, 2, 3])
y = np.array([2, 4, 6])

# Covariance formulas
# Population covariance (ddof=0)
cov_pop = np.cov(x, y, ddof=0)[0, 1]
# Sample covariance (ddof=1)
cov_sample = np.cov(x, y, ddof=1)[0, 1]
# Pearson correlation r
pearson_r = np.corrcoef(x, y)[0, 1]

print("=== Part 1: x = [1, 2, 3], y = [2, 4, 6] ===")
print(f"Population Covariance: {cov_pop:.4f}")
print(f"Sample Covariance: {cov_sample:.4f}")
print(f"Pearson correlation r: {pearson_r:.4f}")
print()

# 2. Replace y with [6, 4, 2]
y_rev = np.array([6, 4, 2])
cov_pop_rev = np.cov(x, y_rev, ddof=0)[0, 1]
cov_sample_rev = np.cov(x, y_rev, ddof=1)[0, 1]
pearson_r_rev = np.corrcoef(x, y_rev)[0, 1]

print("=== Part 2: x = [1, 2, 3], y = [6, 4, 2] ===")
print(f"Population Covariance (reversed): {cov_pop_rev:.4f}")
print(f"Sample Covariance (reversed): {cov_sample_rev:.4f}")
print(f"Pearson correlation r (reversed): {pearson_r_rev:.4f}")
print("Explanation: The sign changes to negative because as x increases, y now decreases.")
print("They move in opposite directions from their respective means, resulting in negative deviation products.")
print()

# 3. Scale units: Express hours (x) in minutes (x * 60)
x_minutes = x * 60
cov_pop_scaled = np.cov(x_minutes, y, ddof=0)[0, 1]
cov_sample_scaled = np.cov(x_minutes, y, ddof=1)[0, 1]
pearson_r_scaled = np.corrcoef(x_minutes, y)[0, 1]

print("=== Part 3: Scaling x to minutes (x * 60) ===")
print(f"New x (minutes): {x_minutes}")
print(f"Scaled Population Covariance: {cov_pop_scaled:.4f} (Original: {cov_pop:.4f})")
print(f"Scaled Sample Covariance: {cov_sample_scaled:.4f} (Original: {cov_sample:.4f})")
print(f"Scaled Pearson correlation r: {pearson_r_scaled:.4f} (Original: {pearson_r:.4f})")
print("Explanation: Covariance changes proportionally to the scaling factor (multiplied by 60) because it depends on the raw units.")
print("Pearson r remains exactly the same because it divides covariance by the standard deviations, canceling out the scale factor.")
```

```python
# Simulation and Empirical probability
# Simulation generates the observations from an assumed the model
# 10,000 groups containing the five independents trails p =0.6
# 0 and 5
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)
sample = rng.binomial(n=5,p=0.6,size=10000)
print(sample)

k = np.arange(6)
print(k)

counts = np.bincount(sample,minlength=6)
print(counts)

empirical = counts / len(sample)
print("Empirical:", empirical)

therotical = stats.binom.pmf(k,n=5,p=0.6)
print("Theoretical:", therotical)

plt.bar(k,empirical,label="Simulation")
plt.scatter(k,therotical,marker="x",color="black",label="Theoretical PMF",zorder=3)
plt.legend()
plt.show()
```
